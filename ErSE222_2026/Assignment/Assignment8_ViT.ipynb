## Assignment: Vision Transformer vs Traditional Deep Learning in Geoscience

### Objective
Replace a traditional deep learning model (e.g., CNN, U-Net, RNN) used in a previous geoscience application from this course with a Vision Transformer (ViT), and compare their performance.

### Tasks
1. Select one geoscience application covered in the course.
2. Reproduce or reuse the baseline model (traditional DL approach).
3. Implement a Vision Transformer model:
   - Use the provided ViT script from the lab folder as a reference.
   - Adapt it to your selected dataset and task.
4. Train both models under similar conditions.
5. Evaluate and compare results using appropriate metrics (e.g., accuracy, F1-score, RMSE, IoU).
